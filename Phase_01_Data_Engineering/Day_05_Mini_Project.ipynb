```python
import sqlite3
conn = sqlite3.connect('student_analytics.db')
cursor = conn.cursor()
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS students (
    student_id INTEGER PRIMARY KEY,
    name TEXT,
    gender TEXT,
    department TEXT,
    math_score INTEGER,
    programming_score INTEGER,
    attendance REAL,
    total_score INTEGER
);
""")
conn.commit()
```

```python
cursor.execute("""
INSERT OR IGNORE INTO students
VALUES
(1,'Arun','Male','CSE',85,90,92,175)
""")
conn.commit()
```

```python
students = [
  (2,'Priya','Female','CSE',88,95,90,183),
  (3,'Karthik','Male','IT',75,80,85,155),
  (4,'Deepika','Female','ECE',92,89,96,181),
  (5,'Rahul','Male','CSE',70,72,80,142)
  ]

cursor.executemany(
  "INSERT OR IGNORE INTO students VALUES (?,?,?,?,?,?,?,?)",
  students
)
conn.commit()
```
